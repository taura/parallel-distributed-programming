# 2. How to work on these notebooks
* The exercises almost never ask you to write a program from scratch. They ask you to change one line and find out what happens. That is the point --- the fastest way to learn what a language construct means is to break it on purpose and read the error message
* Each exercise comes with a starting file (`..._ex.cc`) already containing `main` and the surrounding scaffolding, with comments where your change goes. Edit that copy and leave the demonstration program above it intact, so you can compare

* The programs in this primer are tiny and serial, so they are compiled _and_ run with `%%bash`, on the login node. In the rest of the course you will run programs with `%%sub`, which submits them to a compute node
* `-Wall` asks the compiler to warn about suspicious code. Read the warnings; in C++ a warning is often a bug
