#  C++ Primer : Functions
# 1. Writing a function
* Here is the C++ way of writing a function

```
double average(double a, double b) {
  return (a + b) / 2;
}
```

* Compared with Python's `def average(a, b):`
  * the type of every parameter, and the type of the return value, must be written down
  * the compiler checks every call against them. Calling `average(1, 2, 3)` or `average("x", "y")` is an error caught at compile time, not a surprise at run time
  * a function that returns nothing has the return type `void`
* The basic types you will use in this course:
  * `int` (32 bits), `long` (64 bits)
  * `double` (64-bit floating point), `float` (32 bits)
  * `char`, `bool`
* Arithmetic follows the types. `7 / 2` is `3` because both operands are `int`; `7.0 / 2` is `3.5`. This bites everyone once

# 2. Order matters
* The compiler reads your file from top to bottom and must already know a function when it sees a call to it
* So either define the function before its first use, or put a _declaration_ (the first line followed by `;`) near the top and the definition later

```
double average(double a, double b);   // declaration: "it exists, and this is its type"
```

* This is why C++ files tend to be written bottom-up, with `main` last


````python
%%writefile func.cc
#include <stdio.h>

double average(double a, double b) {
  return (a + b) / 2;
}

int main() {
  printf("average(1, 2) = %f\n", average(1, 2));
  printf("average(3, 4) = %f\n", average(3, 4));
  return 0;
}
````
