#  C++ Primer : Operator Overloading
# 1. Giving `+` a meaning for your own type
* For a type of your own, `a + b` means nothing until you say what it means. You say it by defining a function with a special name

```
vec operator+(vec a, vec b) {
  vec c = { a.x + b.x, a.y + b.y };
  return c;
}
```

* `p + q` is then just a call to that function. There is no magic: an operator is a function with an unusual spelling, chosen and type-checked at compile time like any other
* Most operators can be given a meaning this way: `+ - * / == < [] ()` and more. `+` and `*` on a vector or matrix type is the usual reason to do it
* Two forms, both common:
  * a free function, `vec operator+(vec a, vec b)`, as above
  * a member function, `vec operator+(vec b)` written inside the struct, where the left operand is the object itself
* The rules of the language do not change: `a + b * c` still multiplies first, whatever you made `*` do. You choose the meaning of an operator, never its precedence
* The usual warning applies: define an operator only when the meaning is obvious to the reader. `+` on two vectors is obvious; `+` meaning "append to a log file" is a crime


````python
%%writefile op.cc
#include <stdio.h>
#include <math.h>

struct vec {
  double x;
  double y;
};

vec operator+(vec a, vec b) {
  vec c = { a.x + b.x, a.y + b.y };
  return c;
}

/* scalar times vector */
vec operator*(double k, vec a) {
  vec c = { k * a.x, k * a.y };
  return c;
}

void print(const char * name, vec a) {
  printf("%s = (%f, %f)\n", name, a.x, a.y);
}

int main() {
  vec p = { 1.0, 2.0 };
  vec q = { 10.0, 20.0 };

  print("p", p);
  print("q", q);
  print("p + q", p + q);
  print("2 * p", 2 * p);
  print("p + 2 * q", p + 2 * q);
  return 0;
}
````
