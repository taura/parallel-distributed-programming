#  C++ Primer : Command Line Arguments
# 1. `argc` and `argv`
* `main` may be written with two parameters

```
int main(int argc, char ** argv) {
  ...
}
```

* `argv` is an array of strings and `argc` is how many there are --- the same thing Python calls `sys.argv`
* `argv[0]` is the name of the program itself, so the first real argument is `argv[1]` and the last is `argv[argc - 1]`
* The arguments are always _strings_, even when they look like numbers. To get a number out of one:
  * `atoi(s)` : string to `int`
  * `atol(s)` : string to `long`
  * `atof(s)` : string to `double`
  * they come from `<stdlib.h>`, and they quietly return 0 for anything they cannot parse
* `char ** argv` is a pointer to pointers. You can ignore that for now and read it as "array of strings"

# 2. Giving every argument a default
* A program is much easier to experiment with if every argument is optional. The usual idiom is a conditional expression:

```
long n = (argc > 1 ? atol(argv[1]) : 1000);
```

* `(c ? a : b)` is C++'s conditional expression: `a` if `c` is true, `b` otherwise. So this reads "use the first argument if there is one, otherwise 1000"
* Nearly every program in this course is written this way, so that you can change a problem size on the command line without editing and recompiling


````python
%%writefile args.cc
#include <stdio.h>
#include <stdlib.h>

int main(int argc, char ** argv) {
  for (int i = 0; i < argc; i++) {
    printf("argv[%d] = %s\n", i, argv[i]);
  }
  long n    = (argc > 1 ? atol(argv[1]) : 1000);
  double x  = (argc > 2 ? atof(argv[2]) : 0.5);
  printf("n = %ld, x = %f\n", n, x);
  return 0;
}
````
