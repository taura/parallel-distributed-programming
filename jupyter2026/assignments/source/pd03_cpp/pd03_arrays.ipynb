#  C++ Primer : Arrays
# 1. Declaring an array
* A C++ array is a fixed number of elements of the same type, laid out next to each other in memory

```
double a[5];          // five doubles, values unspecified
long   b[3] = { 10, 20, 30 };
double c[100] = { 0 };  // first element 0, the rest filled with 0 too
```

* Indices start at 0, so the valid indices of `a[5]` are 0, 1, 2, 3 and 4
* The size must be known when the array is declared. `double a[n]` with a variable `n` is not standard C++ (although our compilers accept it). To decide the size at run time, use the heap (see the memory notebook)
* `sizeof(a)` gives the size of the whole array _in bytes_, so `sizeof(a) / sizeof(a[0])` is the number of elements
* This is nothing like a Python list. There is no `append`, no slicing, no negative indexing, and above all

# 2. There is no bounds checking
* `a[10]` on a five-element array is not an error the language reports. It reads or writes whatever happens to sit at that address
* The program may print garbage, may work by luck, may corrupt another variable, or may crash. Which of these happens can change when you recompile with different options
* Keeping indices in range is entirely your job. This is the single most common source of bugs in C and C++


````python
%%writefile array.cc
#include <stdio.h>

/* sum of the first n elements of a */
double sum(double a[], long n) {
  double s = 0;
  for (long i = 0; i < n; i++) {
    s += a[i];
  }
  return s;
}

int main() {
  double a[5] = { 1, 2, 3, 4, 5 };
  long n = sizeof(a) / sizeof(a[0]);
  printf("n = %ld\n", n);
  for (long i = 0; i < n; i++) {
    printf("a[%ld] = %f\n", i, a[i]);
  }
  printf("sum = %f\n", sum(a, n));
  return 0;
}
````

* Note the parameter `double a[]` of `sum`, with no size, and the separate parameter `n`
* An array passed to a function does not carry its length. What the function receives is only the address of the first element, which is why `sizeof` inside `sum` would _not_ give you the size of the array. Passing the length alongside the array is the standard idiom and you will see it everywhere in this course
