#  C++ Primer : Parameter Passing
# 1. Three ways to pass a parameter
* By default C++ passes _by value_: the function gets a copy, and whatever it does to the parameter is invisible to the caller
* When a function must change the caller's variable, you have two ways to say so

| written | at the call | inside the function | changes the caller's variable |
|---|---|---|---|
| `void f(int x)` | `f(a)` | `x` | no |
| `void f(int & x)` | `f(a)` | `x` | yes |
| `void f(int * p)` | `f(&a)` | `* p` | yes |

## 1-1. By value : `int x`
* The copy is what you get. Cheap for an `int`, not so cheap for a large `struct`

## 1-2. By reference : `int & x`
* `int &` is read "reference to `int`". The parameter becomes another name for the caller's variable
* Nothing changes at the call site: you still write `f(a)`. This is convenient and also a little dangerous, because you cannot tell from the call that `a` may be modified
* `const vec & v` is the usual way to pass a large `struct` without copying it while promising not to change it

## 1-3. By pointer : `int * p`
* `int *` is "address of an `int`". `&a` takes the address of `a`, and `* p` is the value stored at that address
* The call site says `f(&a)`, which makes it visible that something is being handed over by address
* A pointer can also be null or be made to point elsewhere, which a reference cannot

* Arrays are a special case: an array parameter is always passed as the address of its first element, so a function always sees the caller's array itself, never a copy. That is why `sum(a, n)` could not copy a million elements by accident, and also why a function _can_ modify the array it is given


```python
%%writefile pass.cc
#include <stdio.h>

void by_value(int x) {
  x = 100;
}

void by_reference(int & x) {
  x = 200;
}

void by_pointer(int * p) {
  * p = 300;
}

int main() {
  int a = 0;

  a = 0; by_value(a);      printf("after by_value     : a = %d\n", a);
  a = 0; by_reference(a);  printf("after by_reference : a = %d\n", a);
  a = 0; by_pointer(&a);   printf("after by_pointer   : a = %d\n", a);
  return 0;
}
```
