#  C++ Primer : Hello World
# 1. The shape of a C++ program
* Here is a simple but complete C++ program

```
#include <stdio.h>

int main() {
  printf("hello world\n");
  return 0;
}
```

* Things to notice, compared with Python:
  * `#include <stdio.h>` makes the names declared in the header `stdio.h` (here, `printf`) usable. It is roughly `import`, but it literally pastes the header's text into your file before compiling
  * you may also see `#include <cstdio>`, the C++ spelling of the same header, which puts the names in the `std` namespace (`std::printf`). This course uses the plain `.h` names
  * execution starts at the function named `main`.
  * `int main()` says `main` returns an `int`. What the `main` returns becomes the exit status of the program. `return 0;` means "finished normally". Any non-zero value means failure
  * every statement ends with `;`
  * `{ ... }` groups statements. Indentation means nothing to the compiler (it means everything to the reader)
  * `printf` does not append a newline; `\n` does
* `printf("format", ...)` takes a format string in which `%d`, `%f` and `%s` are replaced by the remaining arguments
  * `%d` : `int`
  * `%ld` : `long`
  * `%f` : `double` (`%e` for exponent notation, `%g` for whichever is shorter)
  * `%s` : string
  * `%c` : a single character
  * `%%` : a literal `%`
* C++ also has `std::cout << "hello world" << std::endl;`. This course uses `printf`, because the format string makes it easy to control the output

* Save the program
