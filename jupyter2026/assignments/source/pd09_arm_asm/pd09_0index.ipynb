# 4. Assembly code in a nutshell
* A CPU executes _instructions_ (machine code), each of which does a very small job, such as "add two numbers", "read a value from memory", or "jump to another instruction"
* Instructions are binary data; _assembly code_ is their human-readable text form, one instruction per line
* A compiler translates a C/C++ program into instructions; with the `-S` option, it writes them out as assembly code (`xxx.c` becomes `xxx.s`) instead of producing an executable
* Different CPUs have different instruction sets. Miyabi's CPU (NVIDIA Grace) implements the instruction set of _Arm_ (more precisely, the 64-bit Arm architecture called _AArch64_ or _ARM64_), which is what we see here. Most laptops and servers use Intel/AMD CPUs (x86-64), whose instructions look different, but the ideas are the same

# 5. The compiler
* We use the same compiler as the rest of the course, `clang` (for C) and `clang++` (for C++)
* Every cell that compiles a program starts with this line, which makes them available
```
module purge && module load nvidia/25.9 cuda/12.9 && PATH=/work/gt81/share/env/llvm/bin:$PATH
```
* We compile with `-O1`, a modest level of optimization
  * without optimization (`-O0`), the compiler translates every C statement literally and keeps all variables in memory, which makes the code long and hard to read
  * with higher optimization (`-O2`, `-O3`), the compiler starts using _SIMD instructions_ for loops, which we will learn later (pd07_simd)
* The programs are tiny, so they are compiled (and occasionally run) with `%%bash`, on the login node
